```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
df.head()
```

```python
print(df.shape)
print(df.info())
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['total_bill'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='day', y='tip', data=df, palette='viridis', errorbar=None)
plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df, palette='Set1')
plt.title('Total Bill vs Tip by Gender')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='day', y='total_bill', hue='time', data=df, palette='pastel')
plt.title('Total Bill Distribution by Day and Time')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')
plt.legend(title='Time')
plt.show()
```

```python
plt.figure(figsize=(6,5))
numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(6,6))
df['smoker'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#66b3ff','#ff9999'], startangle=90)
plt.title('Smoker vs Non-Smoker Proportion')
plt.ylabel('')
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14,10))

sns.histplot(df['total_bill'], bins=20, kde=True, color='skyblue', ax=axes[0,0])
axes[0,0].set_title('Total Bill Distribution')

sns.barplot(x='day', y='tip', data=df, palette='viridis', errorbar=None, ax=axes[0,1])
axes[0,1].set_title('Average Tip by Day')

sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df, palette='Set1', ax=axes[1,0])
axes[1,0].set_title('Total Bill vs Tip')

sns.boxplot(x='day', y='total_bill', hue='time', data=df, palette='pastel', ax=axes[1,1])
axes[1,1].set_title('Total Bill by Day & Time')

plt.suptitle('Tips Dataset - Data Visualization Dashboard', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```
